```python
import pandas as pd
import zipfile
import os

# Define the path to the zip file and the CSV file name
zip_file_path = '/content/archive.zip'
csv_file_name = 'student_exam_performance.csv'

# Check if the zip file exists and extract its contents
if os.path.exists(zip_file_path):
    print(f"Extracting {zip_file_path}...")
    with zipfile.ZipFile(zip_file_path, 'r') as zip_ref:
        # Extract all contents to the current directory
        zip_ref.extractall('.')
    print("Extraction complete.")
else:
    print(f"Error: Zip file not found at {zip_file_path}. Please ensure it is uploaded.")

# Now, attempt to read the CSV file, assuming it's extracted to the current directory.
df = pd.read_csv(csv_file_name)

print("Dataset Shape:", df.shape)

print("\n--- First 5 Rows ---")
display(df.head())

print("\n--- Dataset Info ---")
df.info()
```
